```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import mnist
from tensorflow.keras.layers import *
from tensorflow.keras.models import Model
from tensorflow.keras import regularizers
```

```python
(x_train,_),(x_test,_) = mnist.load_data()

x_train = x_train.astype("float32")/255.0

x_test = x_test.astype("float32")/255.0

x_train = x_train.reshape(-1,784)
x_test = x_test.reshape(-1,784)

print(x_train.shape)

print(x_test.shape)
```

```python
from threading import active_count
inp = Input(shape=(784,))

x = Dense (
    128,
    activation='relu',
    activity_regularizer=regularizers.l1(1e-4)
)(inp)


out = Dense(784,activation='sigmoid')(x)

model_l1 = Model(inp,out)

model_l1.compile(
    optimizer='adam',
    loss='binary_crossentropy'
)
```

```python
model_l1.fit(x_train,x_train,epochs=5,batch_size=256,shuffle=True,validation_data=(x_test,x_test))
```

```python
inp = Input(shape=(784,))

x = Dense(
    128,
    activation='relu',
    activity_regularizer=regularizers.l1_l2(l1=1e-4,l2=1e-4)
)(inp)

out = Dense(784,activation='sigmoid')(x)

model_l2 = Model(inp,out)

model_l2.compile(
    optimizer='adam',
    loss='binary_crossentropy'
)
```

```python
model_l2.fit (
    x_train,x_train,epochs=5,batch_size=256,shuffle=True,validation_data=(x_test,x_test)
)
```

```python
inp = Input(shape=(784,))

x = Dense(
    256,
    activation='relu',
    activity_regularizer=regularizers.l2(1e-4)
)(inp)

x = Dropout(0.3)(x)

x = Dense(
    128,
    activation='relu',
    activity_regularizer=regularizers.l2(1e-4)
)(x)

x = Dropout(0.3)(x)

out = Dense(784,activation='sigmoid')(x)

model_dropout = Model(inp,out)

model_dropout.compile(
    optimizer='adam',
    loss='binary_crossentropy'
)
```

```python
model_dropout.fit(
    x_train,x_train,epochs=5,batch_size=256,shuffle=True,validation_data=(x_test,x_test)
)
```

```python
inp = Input(shape=(784,))

x = Dense(
    256,
    activation='relu',
    activity_regularizer=regularizers.l2(1e-4)
)(inp)

x = BatchNormalization()(x)

x = Dense(
    128,
    activation='relu',
    activity_regularizer=regularizers.l2(1e-4)
)(inp)

x = BatchNormalization()(x)

out = Dense(784,activation='sigmoid')(x)

model_batch = Model(inp,out)

model_batch.compile(
    optimizer='adam',
    loss='binary_crossentropy'
)
```

```python
model_batch.fit(
    x_train,x_train,epochs=5,batch_size=256,shuffle=True,validation_data=(x_test,x_test)
)
#
```

```python
early_stop = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=10,
    verbose=1,
    mode='min',
    restore_best_weights=True
)
```

```python
inp = Input(shape=(784,))

x = Dense(
    256,
    activation='relu',
    activity_regularizer=regularizers.l2(1e-4)
)(inp)

x = Dense(
    128,
    activation='relu',
    activity_regularizer=regularizers.l2(1e-4)
)(x)

out = Dense(784,activation='sigmoid')(x)

model_es = Model(inp,out)

model_es.compile(
    optimizer='adam',
    loss='binary_crossentropy'
)
```

```python
model_es.fit(x_train,x_train,epochs=5,batch_size=256,shuffle=True,validation_data=(x_test,x_test),callbacks=[early_stop])
```
